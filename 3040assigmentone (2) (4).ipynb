```python
import numpy as  np
import matplotlib.pyplot as plt

# the Euclidean distance  function
def euclidean(x, y):
    return np.sqrt(np.sum((x - y) ** 2))

# KNN imputation function  (same as in que 2)
def knn_impute(matrix_with_missing, K=3):
    matrix = matrix_with_missing.copy()
    n, p = matrix.shape
    obs = ~np.isnan(matrix)

    # Loop through each row (i) to find missing values
    for i in range(n):
        missing_cols =  np.where(~obs[i])[0]

        if missing_cols.size == 0:
            continue

        for j  in missing_cols:
            candidates = []

            # Loop , nearest neighbors
            for k in range(n):
                if k ==  i or not obs[k, j]:
                    continue

                shared = obs[i] & obs[k]
                shared[j] = False

                if not  np.any(shared):
                    continue

                # C distance btween rows
                d_ik = euclidean(matrix[i, shared], matrix[k, shared])
                candidates.append((d_ik, matrix[k, j]))

            # If we have at least one candiat, the avg the K nearest neighbors
            if candidates:
                candidates.sort(key=lambda t: t[0])
                topK = candidates[:min(K, len(candidates))]
                neighbor_vals = np.array([val for (_, val) in topK], dtype=float)
                matrix[i, j] =  neighbor_vals.mean()
            else:
                matrix[i, j] = np.nanmean(matrix[:, j])
    return matrix

# Function to comp the imputation error (Euclidean distance)
def compute_imputation_error(matrix_full, matrix_with_missing, nan_positions):
    v = np.array([matrix_full[i, j] for  (i, j) in nan_positions])
    e = np.array([matrix_with_missing[i, j] for (i, j) in nan_positions])
    error = np.sqrt(np.sum((v - e) ** 2))
    return error

def q3_run():
    # will  the matrix with missing values ( auestion 1)
    matrix_full, matrix_with_missing = generate_matrix_with_nans()

    # Get the positions of missing  values
    nan_positions = [(i, j) for i in range(1000) for j in  range(5) if np.isnan(matrix_with_missing[i, j])]

    #  List will to store errors for diff values of K
    Ks = list(range(1, 11))
    errors = []

    # Loop over K values from 1 to 10 and compute the  error for each K
    for K in Ks:
        matrix_imputed =  knn_impute(matrix_with_missing, K)
        error = compute_imputation_error(matrix_full, matrix_imputed, nan_positions)
        errors.append(error)
        print(f"Q3 → K={K}:  d(v, e)={error:.6f}")

    best_K = Ks[np.argmin(errors)]
    print(f"Q3 →   Best K: {best_K}")

    # Plot
    plt.plot(Ks, errors, marker='o')
    plt.xlabel("K (Number of Neighbors)")
    plt.ylabel ("Imputation  Error (Euclidean distance)")
    plt.title("Q3  – KNN Imputation Error vs K")
    plt.grid(True)
    plt.savefig("q3_knn_error_vs_k.png",  bbox_inches="tight")
    plt.show()

q3_run()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import  seaborn as sns
import scipy.stats as stats

bike_data = pd.read_csv('bike_data.csv')

print(bike_data.head())

# EDA Measures  for temp and Bike
for col in ['temperature', 'bike_counts']:
    print(f"\n{col.capitalize()}:")

    print(f"Mean: {bike_data[col].mean():.2f}")

    print(f"Median: {bike_data[col].median():.2f}")

    mode = bike_data[col].mode()[0]
    print(f"Mode: {mode:.2f}")

    print(f"Variance: {bike_data[col].var():.2f}")

    print(f"Standard Deviation: {bike_data[col].std():.2f}")

# Boxplot for temp and  bike_counts
plt.figure(figsize=(12, 6))
sns.boxplot(data=bike_data[['temperature', 'bike_counts']])
plt.title("Boxplot of Temperature and Bike Counts")
plt.show()

# Histogram  for bike and temp
plt.figure(figsize=(12, 6))
bike_data[['temperature', 'bike_counts']].hist(bins=20, edgecolor='black')
plt.suptitle("Histogram of Temperature and Bike Counts")
plt.show()

# QQ-Plot for bike and temp
plt.figure(figsize=(12, 6))

# QQ-plot for temo
plt.subplot(1, 2, 1)
stats.probplot(bike_data['temperature'].dropna(), dist="norm", plot=plt)
plt.title("QQ-Plot: Temperature")

# QQ-pl for  bike_counts
plt.subplot(1,  2, 2)
stats.probplot(bike_data['bike_counts'].dropna(), dist="norm", plot=plt)
plt.title("QQ-Plot: Bike Counts")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import  matplotlib.pyplot as plt
import seaborn as sns

bike_data =  pd.read_csv('bike_data.csv')

print(bike_data.head())

# make  a scatter plot of bike counts vs temp
plt.figure(figsize=(8, 6))
sns.scatterplot(x='temp', y='bike_counts', data=bike_data)
plt.title("Scatter Plot: Bike Counts vs temp")
plt.xlabel("temp")
plt.ylabel("Bike Counts")
plt.grid(True)
plt.show()

# have a compute the Pearson correlation using Pandas
pearson_corr_pandas = bike_data['temp'].corr(bike_data['bike_counts'])
print(f"Pearson Correlation (using Pandas): {pearson_corr_pandas:.4f}")

def pearson_correlation(x, y):
    x = x.dropna()
    y = y.dropna()

    mean_x = x.mean()
    mean_y = y.mean()

    numerator = np.sum((x - mean_x) * (y - mean_y))
    denominator = np.sqrt(np.sum((x - mean_x)**2) * np.sum((y - mean_y)**2))

    return numerator / denominator

pearson_corr_scratch = pearson_correlation(bike_data['temperature'], bike_data['bike_counts'])
print(f"Pearson Correlation (from scratch): {pearson_corr_scratch:.4f}")
```
